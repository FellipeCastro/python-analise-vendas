# **Análise de Vendas de uma Loja**
*O objetivo dessa análise é compreender pontos importantes de vendas de uma loja, bem como:*
* Qual produto mais vende?
* Qual categoria dá mais lucro?
* Quais meses têm mais vendas?
* Qual região vende mais?



## **Carregando dados**

```python
import pandas as pd

df = pd.read_csv("train.csv")
df.head()
```

## **Analisando dados**

```python
df.info()
```

```python
df.describe()
```

## **Respondendo perguntas**

### **Produto mais vendido**



```python
# quantidade vendida de cada produto
top_produtos = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False).head()
top_produtos
```

### **Categoria mais lucrativa**

```python
# total de lucro de cada categoria
lucro_categoria = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
lucro_categoria
```

### **Vendas por mês**

```python
# formatando a data
df["Order Date"] = pd.to_datetime(df["Order Date"])

# criando coluna de mês
df["Month"] = df["Order Date"].dt.month

# vendas por mês
vendas_mes = df.groupby("Month")["Sales"].sum()
vendas_mes
```

### **Região mais lucrativa**

```python
regiao_mais_lucrativa = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)
regiao_mais_lucrativa
```

## **Gerando relatorio Excel**

```python
with pd.ExcelWriter("relatorio.xlsx") as writer:
  df.to_excel(writer, sheet_name="Dados Brutos", index=False)
  top_produtos.to_excel(writer, sheet_name="Top Produtos")
  lucro_categoria.to_excel(writer, sheet_name="Lucro por categoria")
  vendas_mes.to_excel(writer, sheet_name="Vendas por mês")
  regiao_mais_lucrativa.to_excel(writer, sheet_name="Região mais lucrativa")
```
